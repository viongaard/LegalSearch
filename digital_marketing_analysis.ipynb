# Анализ данных датасета "Digital Marketing Performance Dataset"

Загрузка датасета

```python
import kagglehub
import pandas as pd
import os

# Download latest version
path = kagglehub.dataset_download("alinaboulsi/digital-marketing-performance-dataset")

print("Path to dataset files:", path)

data_dict = pd.read_csv(os.path.join(path, "data_dictionary.csv"))
df = pd.read_csv(os.path.join(path, "digital_marketing_dataset_30k.csv"))
```

Информация о датасете (пропусков нет, что удивительно)

```python
# Общее описание
print(df.shape)
print(df.info(), '/n')
print(df.head(), '/n')
print(df.describe(), '/n')

# Пропуски
import seaborn as sns
import matplotlib.pyplot as plt

print(df.isnull().sum())
plt.figure(figsize=(12, 6))
sns.heatmap(df.isnull(), cbar=True, yticklabels=False, cmap='viridis')
plt.title('Визуализация пропусков в данных')
plt.show()
```
